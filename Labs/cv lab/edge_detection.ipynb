```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image in grayscale
image = cv2.imread('image.png', cv2.IMREAD_GRAYSCALE)

# Convert image to float32 for gradient calculation
image_float = np.float32(image)

# Prewitt kernels
prewitt_kernel_x = np.array([[1, 0, -1],
                             [1, 0, -1],
                             [1, 0, -1]])

prewitt_kernel_y = np.array([[1, 1, 1],
                             [0, 0, 0],
                             [-1, -1, -1]])

# Apply Prewitt filter
prewitt_x = cv2.filter2D(image_float, -1, prewitt_kernel_x)
prewitt_y = cv2.filter2D(image_float, -1, prewitt_kernel_y)

# Calculate magnitude of gradients
prewitt_edge = cv2.magnitude(prewitt_x, prewitt_y)

# Show result
plt.imshow(prewitt_edge, cmap='gray')
plt.title("Prewitt Edge Detection")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image in grayscale
image = cv2.imread('image.png', cv2.IMREAD_GRAYSCALE)

# Convert image to float32 for gradient calculation
image_float = np.float32(image)

# Roberts Cross kernels (2x2)
roberts_kernel_x = np.array([[1, 0],
                              [0, -1]])

roberts_kernel_y = np.array([[0, 1],
                              [-1, 0]])

# Apply Roberts filter (use border handling to avoid issues with 2x2 kernels)
roberts_x = cv2.filter2D(image_float, -1, roberts_kernel_x)
roberts_y = cv2.filter2D(image_float, -1, roberts_kernel_y)

# Calculate magnitude of gradients
roberts_edge = cv2.magnitude(roberts_x, roberts_y)

# Show result
plt.imshow(roberts_edge, cmap='gray')
plt.title("Roberts Edge Detection")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image in grayscale
image = cv2.imread('image.png', cv2.IMREAD_GRAYSCALE)

# Apply Canny edge detection
# The two threshold values are chosen empirically, lower and upper thresholds.
canny_edges = cv2.Canny(image, 100, 200)

# Show result
plt.imshow(canny_edges, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image in grayscale
image = cv2.imread('image.png', cv2.IMREAD_GRAYSCALE)

# Apply Gaussian filter (Blurring) first
gaussian_blur = cv2.GaussianBlur(image, (5, 5), 0)

# Apply Laplacian of Gaussian (LoG)
laplacian = cv2.Laplacian(gaussian_blur, cv2.CV_64F)

# Convert to absolute values and convert back to uint8
laplacian_abs = cv2.convertScaleAbs(laplacian)

# Show result
plt.imshow(laplacian_abs, cmap='gray')
plt.title("Laplacian of Gaussian (LoG)")
plt.axis('off')
plt.show()
```
