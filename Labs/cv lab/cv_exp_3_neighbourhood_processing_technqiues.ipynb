```python
#exp 3

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image in grayscale
img = cv2.imread('/content/sesko.jpg', cv2.IMREAD_GRAYSCALE)

# Define Roberts cross operator kernels
roberts_cross_v = np.array([[1, 0],
                            [0, -1]], dtype=np.float32)

roberts_cross_h = np.array([[0, 1],
                            [-1, 0]], dtype=np.float32)

# Apply kernels
img_roberts_v = cv2.filter2D(img, -1, roberts_cross_v)
img_roberts_h = cv2.filter2D(img, -1, roberts_cross_h)

# Combine edges
img_roberts = cv2.addWeighted(img_roberts_v, 0.5, img_roberts_h, 0.5, 0)

# Show results
plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Roberts Edge Detection')
plt.imshow(img_roberts, cmap='gray')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Load image in grayscale
img = cv2.imread('/content/sesko.jpg', cv2.IMREAD_GRAYSCALE)

# Define Prewitt kernels
prewitt_x = np.array([[ -1, 0, 1],
                      [ -1, 0, 1],
                      [ -1, 0, 1]], dtype=np.float32)

prewitt_y = np.array([[ 1,  1,  1],
                      [ 0,  0,  0],
                      [-1, -1, -1]], dtype=np.float32)

# Apply kernels
img_prewitt_x = cv2.filter2D(img, -1, prewitt_x)
img_prewitt_y = cv2.filter2D(img, -1, prewitt_y)

# Combine edges
img_prewitt = cv2.addWeighted(img_prewitt_x, 0.5, img_prewitt_y, 0.5, 0)

plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Prewitt Edge Detection')
plt.imshow(img_prewitt, cmap='gray')
plt.axis('off')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image in grayscale
img = cv2.imread('/content/sesko.jpg', cv2.IMREAD_GRAYSCALE)

# Apply Sobel operator
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

# Calculate gradient magnitude
sobel = cv2.magnitude(sobel_x, sobel_y)
sobel = cv2.convertScaleAbs(sobel)

plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Sobel Edge Detection')
plt.imshow(sobel, cmap='gray')
plt.axis('off')

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image in grayscale
img = cv2.imread('/content/sesko.jpg', cv2.IMREAD_GRAYSCALE)

# Apply Laplacian operator
laplacian = cv2.Laplacian(img, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

# Show results
plt.subplot(1, 2, 1)
plt.title('Original')
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Laplacian Edge Detection')
plt.imshow(laplacian, cmap='gray')
plt.axis('off')

plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def add_salt_pepper_noise(image, salt_prob=0.02, pepper_prob=0.02):
    noisy_img = image.copy()
    total_pixels = image.size

    # Salt noise
    num_salt = int(total_pixels * salt_prob)
    coords = [np.random.randint(0, i - 1, num_salt) for i in image.shape]
    noisy_img[coords[0], coords[1]] = 255

    # Pepper noise
    num_pepper = int(total_pixels * pepper_prob)
    coords = [np.random.randint(0, i - 1, num_pepper) for i in image.shape]
    noisy_img[coords[0], coords[1]] = 0

    return noisy_img




    # Load grayscale image
img = cv2.imread('/content/sesko.jpg', cv2.IMREAD_GRAYSCALE)

# Add salt and pepper noise
noisy_img = add_salt_pepper_noise(img, salt_prob=0.02, pepper_prob=0.02)

# Show noisy image
plt.title('Noisy Image (Salt & Pepper)')
plt.imshow(noisy_img, cmap='gray')
plt.axis('off')
plt.show()
```
